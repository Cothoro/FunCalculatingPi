```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Simulation parameters
# In thes visualization program lets assume a screen space of 1000 pixels and 60 FPS
# For this notebook we don't care about drawing size, just the positions.

DIGITS_TO_CALCULATE = 8

TIME_STEP = 1/60 # seconds per frame

BARRIER_X = 0

SMALL_BLOCK_INITIAL_X = 250
SMALL_BLOCK_INITIAL_DX = 0

LARGE_BLOCK_INITIAL_X = 500
LARGE_BLOCK_INITIAL_DX  = -100 # pixels per second

SMALL_BLOCK_MASS = 1
LARGE_BLOCK_MASS = 100**(DIGITS_TO_CALCULATE-1)
```

```python
class Block:
    def __init__(self, x, dx, mass):
        self.x = x
        self.dx = dx
        self.mass = mass
    
    

small_block = Block(SMALL_BLOCK_INITIAL_X, SMALL_BLOCK_INITIAL_DX, SMALL_BLOCK_MASS)
large_block = Block(LARGE_BLOCK_INITIAL_X, LARGE_BLOCK_INITIAL_DX, LARGE_BLOCK_MASS)
```

```python
def time_to_block_collision(small, large):
    """Time until the two blocks collide."""
    relative_velocity = small.dx - large.dx
    if relative_velocity <= 0:
        return float('inf')
    t = (large.x - small.x) / relative_velocity
    return t if t > 0 else float('inf')

def time_to_wall_collision(small):
    """Time until the small block hits the wall at BARRIER_X."""
    if small.dx >= 0:
        return float('inf')
    t = -small.x / small.dx
    return t if t > 0 else float('inf')

def elastic_collision(b1, b2):
    """Apply elastic collision formulas to two blocks."""
    m1, m2 = b1.mass, b2.mass
    v1, v2 = b1.dx, b2.dx
    total = m1 + m2
    b1.dx = ((m1 - m2) * v1 + 2 * m2 * v2) / total
    b2.dx = ((m2 - m1) * v2 + 2 * m1 * v1) / total

def simulate():
    small = Block(SMALL_BLOCK_INITIAL_X, SMALL_BLOCK_INITIAL_DX, SMALL_BLOCK_MASS)
    large = Block(LARGE_BLOCK_INITIAL_X, LARGE_BLOCK_INITIAL_DX, LARGE_BLOCK_MASS)
    
    collisions = 0
    
    while True:
        t_blocks = time_to_block_collision(small, large)
        t_wall = time_to_wall_collision(small)
        
        t = min(t_blocks, t_wall)
        if t == float('inf'):
            break
        
        small.x += small.dx * t
        large.x += large.dx * t
        
        if t_wall < t_blocks:
            small.x = 0
            small.dx = -small.dx
        else:
            elastic_collision(small, large)
        
        collisions += 1
    
    return collisions

count = simulate()
print(f"Number of collisions: {count}")
print(f"Expected (pi digits):  {str(np.pi)[:DIGITS_TO_CALCULATE+1]}")
print(f"Digits of pi: {np.pi:.10f}")
```
